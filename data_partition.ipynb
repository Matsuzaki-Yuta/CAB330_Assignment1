```python
import pandas as pd
df_raw = pd.read_csv('data/clean_data.csv')
```

```python
from sklearn.model_selection import train_test_split

target = "is_canceled" # replace with the target_variable name.

# ---------------------------------------------------------
# Step 1: Reserve 9% for the three individual test datasets
# ---------------------------------------------------------
group_data, test_pool = train_test_split(
    df_raw,
    test_size=0.09,
    random_state=42,
    stratify=df_raw[target]
)

# --------------------------------------------------
# Step 2: Take one-third of the test pool as Test A
# Test A = approximately 3% of the complete dataset
# --------------------------------------------------
test_a, remaining_test_pool = train_test_split(
    test_pool,
    test_size=2 / 3,
    random_state=43,
    stratify=test_pool[target]
)

# -------------------------------------------------------------
# Step 3: Divide the remaining pool equally into Tests B and C
# -------------------------------------------------------------
test_b, test_c = train_test_split(
    remaining_test_pool,
    test_size=0.50,
    random_state=44,
    stratify=remaining_test_pool[target]
)

# Reset row indices
group_data = group_data.reset_index(drop=True)
test_a = test_a.reset_index(drop=True)
test_b = test_b.reset_index(drop=True)
test_c = test_c.reset_index(drop=True)

# Save the datasets
group_data.to_csv(
    "data/clean_group_data.csv",
    index=False
)

test_a.to_csv(
    "data/clean_test_A.csv",
    index=False
)

test_b.to_csv(
    "data/clean_test_B.csv",
    index=False
)

test_c.to_csv(
    "data/clean_test_C.csv",
    index=False
)
```
